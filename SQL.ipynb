1. Explain the fundamental differences between DDL, DML, and DQL
commands in SQL. Provide one example for each type of command.

     - SQL: DDL vs DML vs DQL

1. DDL (Data Definition Language)
Definition:
DDL commands are used to define and modify the structure of database objects like tables.

Examples:
CREATE, ALTER, DROP, TRUNCATE

Example Query:
CREATE TABLE Students (
    ID INT,
    Name VARCHAR(50),
    Age INT
);

--------------------------------------------------

2. DML (Data Manipulation Language)
Definition:
DML commands are used to manipulate data stored inside tables.

Examples:
INSERT, UPDATE, DELETE

Example Query:
INSERT INTO Students (ID, Name, Age)
VALUES (1, 'Rahul', 20);

--------------------------------------------------

3. DQL (Data Query Language)
Definition:
DQL is used to retrieve data from the database.

Main Command:
SELECT

Example Query:
SELECT * FROM Students;

--------------------------------------------------

Summary:
DDL → Structure create/modify
DML → Data insert/update/delete
DQL → Data retrieve (select)

2. What is the purpose of SQL constraints? Name and describe three common types
of constraints, providing a simple scenario where each would be useful.

       SQL Constraints – Purpose and Types

Purpose of SQL Constraints:
SQL constraints are rules applied to table columns to control the type of data that can be stored in a table. They help maintain data accuracy, integrity, and reliability in a database.

--------------------------------------------------

1. NOT NULL Constraint
Description:
Ensures that a column cannot have NULL (empty) values.

Use Case Scenario:
In a Students table, every student must have a name.

Example:
CREATE TABLE Students (
    ID INT,
    Name VARCHAR(50) NOT NULL,
    Age INT
);

Why useful:
It prevents missing important data like student name.

--------------------------------------------------

2. UNIQUE Constraint
Description:
Ensures that all values in a column are different (no duplicates allowed).

Use Case Scenario:
Each student must have a unique email address.

Example:
CREATE TABLE Students (
    ID INT,
    Email VARCHAR(100) UNIQUE
);

Why useful:
It avoids duplicate email entries in the database.

--------------------------------------------------

3. PRIMARY KEY Constraint
Description:
A combination of NOT NULL and UNIQUE. It uniquely identifies each record in a table.

Use Case Scenario:
Student ID should be unique for every student.

Example:
CREATE TABLE Students (
    ID INT PRIMARY KEY,
    Name VARCHAR(50),
    Age INT
);

Why useful:
It ensures every record can be uniquely identified.

--------------------------------------------------

Summary:
NOT NULL → No empty values allowed
UNIQUE → No duplicate values allowed
PRIMARY KEY → Unique + Not Null identifier for each row

3. Explain the difference between LIMIT and OFFSET clauses in SQL. How
would you use them together to retrieve the third page of results, assuming each page
has 10 records?

      - SQL: LIMIT vs OFFSET

Difference:

1. LIMIT
LIMIT is used to restrict the number of rows returned by a query.

Example:
SELECT * FROM Students
LIMIT 10;

→ This returns only 10 records from the table.

--------------------------------------------------

2. OFFSET
OFFSET is used to skip a specific number of rows before starting to return results.

Example:
SELECT * FROM Students
OFFSET 10;

→ This skips the first 10 records and returns the remaining rows.

--------------------------------------------------

Using LIMIT + OFFSET together (Pagination):

To get paginated results, we use both LIMIT and OFFSET.

Formula:
OFFSET = (page number - 1) × page size

Given:
Page size = 10 records
We want: 3rd page

Calculation:
OFFSET = (3 - 1) × 10 = 20

Query:
SELECT * FROM Students
LIMIT 10
OFFSET 20;

--------------------------------------------------

Final Answer:
- LIMIT controls how many rows are shown
- OFFSET controls where to start
- 3rd page (10 per page) → LIMIT 10 OFFSET 20

4. What is a Common Table Expression (CTE) in SQL, and what are its main
benefits? Provide a simple SQL example demonstrating its usage.

        - SQL: Common Table Expression (CTE)

What is a CTE?
A Common Table Expression (CTE) is a temporary named result set that you can define within a SQL query using the WITH clause. It exists only during the execution of that query.

--------------------------------------------------

Main Benefits of CTE:

1. Improves Readability
- Breaks complex queries into simple, logical parts.

2. Reusability within query
- Same result set can be referenced multiple times inside the query.

3. Simplifies complex joins and subqueries
- Makes queries easier to understand and maintain.

4. Useful for recursion
- Can handle hierarchical data (like employee-manager structures).

--------------------------------------------------

Simple Example:

Scenario:
We want to find employees with salary greater than the average salary.

SQL using CTE:

WITH AvgSalary AS (
    SELECT AVG(Salary) AS avg_sal
    FROM Employees
)
SELECT *
FROM Employees
WHERE Salary > (SELECT avg_sal FROM AvgSalary);

--------------------------------------------------

Explanation:
- First, CTE calculates average salary.
- Then main query uses that result to filter employees.

--------------------------------------------------

Final Summary:
CTE = Temporary named query result used to simplify complex SQL and improve readability.

5. Describe the concept of SQL Normalization and its primary goals. Briefly
explain the first three normal forms (1NF, 2NF, 3NF).


Primary Goals of Normalization:

1. Reduce data redundancy (avoid duplicate data)
2. Improve data integrity (correct and consistent data)
3. Make database efficient and easier to maintain
4. Eliminate update, insert, and delete anomalies

--------------------------------------------------

1NF (First Normal Form)

Definition:
A table is in 1NF if:
- Each column contains atomic (indivisible) values
- Each record is unique
- No repeating groups or arrays

Example:
Bad Table:
Student(ID, Name, Subjects)
(1, Rahul, Math, Science)

1NF Table:
Student(ID, Name, Subject)
(1, Rahul, Math)
(1, Rahul, Science)

--------------------------------------------------

2NF (Second Normal Form)

Definition:
A table is in 2NF if:
- It is already in 1NF
- All non-key attributes fully depend on the whole primary key (no partial dependency)

Example:
Bad Table:
StudentCourse(StudentID, CourseID, StudentName)

Problem:
StudentName depends only on StudentID (partial dependency)

2NF Solution:
Student(StudentID, StudentName)
CourseEnrollment(StudentID, CourseID)

--------------------------------------------------

3NF (Third Normal Form)

Definition:
A table is in 3NF if:
- It is in 2NF
- No transitive dependency (non-key attribute should not depend on another non-key attribute)

Example:
Bad Table:
Student(StudentID, Name, DeptID, DeptName)

Problem:
DeptName depends on DeptID, not directly on StudentID

3NF Solution:
Student(StudentID, Name, DeptID)
Department(DeptID, DeptName)

--------------------------------------------------

Final Summary:
1NF → Atomic values, no repeating groups
2NF → No partial dependency
3NF → No transitive dependency

6.  Generate a report showing CustomerName, Email, and the
TotalNumberofOrders for each customer. Include customers who have not placed
any orders, in which case their TotalNumberofOrders should be 0. Order the results
by CustomerName

      

```python
-- Report: CustomerName, Email, TotalNumberOfOrders (including customers with 0 orders)

SELECT
    c.CustomerName,
    c.Email,
    COUNT(o.OrderID) AS TotalNumberOfOrders
FROM Customers c
LEFT JOIN Orders o
    ON c.CustomerID = o.CustomerID
GROUP BY
    c.CustomerName,
    c.Email
ORDER BY
    c.CustomerName;
```

7. Retrieve Product Information with Category: Write a SQL query to

    p.ProductName,
    p.Price,

    c.CategoryName
FROM Products p

    ON p.CategoryID = c.CategoryID
ORDER BY
    c.CategoryName ASC,

8. Write a SQL query that uses a Common Table Expression (CTE) and a
Window Function (specifically ROW_NUMBER() or RANK()) to display the
CategoryName, ProductName, and Price for the top 2 most expensive products in
each CategoryName.

      - -- Top 2 most expensive products in each category using CTE + Window Function

WITH RankedProducts AS (
    SELECT
        c.CategoryName,
        p.ProductName,
        p.Price,
        ROW_NUMBER() OVER (
            PARTITION BY c.CategoryName
            ORDER BY p.Price DESC
        ) AS rn
    FROM Products p
    JOIN Categories c
        ON p.CategoryID = c.CategoryID
)

SELECT
    CategoryName,
    ProductName,
    Price
FROM RankedProducts
WHERE rn <= 2
ORDER BY
    CategoryName,
    Price DESC;

9. You are hired as a data analyst by Sakila Video Rentals, a global movie
rental company. The management team is looking to improve decision-making by
analyzing existing customer, rental, and inventory data.
Using the Sakila database, answer the following business questions to support key strategic
initiatives.
Tasks & Questions:
1. Identify the top 5 customers based on the total amount they’ve spent. Include customer
name, email, and total amount spent.
2. Which 3 movie categories have the highest rental counts? Display the category name
and number of times movies from that category were rented.
3. Calculate how many films are available at each store and how many of those have
never been rented.
4. Show the total revenue per month for the year 2023 to analyze business seasonality.
5. Identify customers who have rented more than 10 times in the last 6 month

-- Sakila Video Rentals: Business Analysis Queries

--------------------------------------------------
-- 1. Top 5 customers by total amount spent
--------------------------------------------------
SELECT
    c.first_name,
    c.last_name,
    c.email,
    SUM(p.amount) AS total_spent
FROM customer c
JOIN payment p
    ON c.customer_id = p.customer_id
GROUP BY
    c.customer_id, c.first_name, c.last_name, c.email
ORDER BY total_spent DESC
LIMIT 5;

--------------------------------------------------
-- 2. Top 3 movie categories by rental count
--------------------------------------------------
SELECT
    cat.name AS category_name,
    COUNT(r.rental_id) AS total_rentals
FROM category cat
JOIN film_category fc
    ON cat.category_id = fc.category_id
JOIN film f
    ON fc.film_id = f.film_id
JOIN inventory i
    ON f.film_id = i.film_id
JOIN rental r
    ON i.inventory_id = r.inventory_id
GROUP BY cat.name
ORDER BY total_rentals DESC
LIMIT 3;

--------------------------------------------------
-- 3. Films available per store + never rented
--------------------------------------------------
SELECT
    i.store_id,
    COUNT(i.inventory_id) AS total_films_available,
    SUM(CASE WHEN r.rental_id IS NULL THEN 1 ELSE 0 END) AS never_rented
FROM inventory i
LEFT JOIN rental r
    ON i.inventory_id = r.inventory_id
GROUP BY i.store_id;

--------------------------------------------------
-- 4. Total revenue per month for year 2023
--------------------------------------------------
SELECT
    DATE_FORMAT(payment_date, '%Y-%m') AS month,
    SUM(amount) AS total_revenue
FROM payment
WHERE YEAR(payment_date) = 2023
GROUP BY DATE_FORMAT(payment_date, '%Y-%m')
ORDER BY month;

--------------------------------------------------
-- 5. Customers who rented more than 10 times in last 6 months
--------------------------------------------------
SELECT
    c.customer_id,
    c.first_name,
    c.last_name,
    COUNT(r.rental_id) AS rental_count
FROM customer c
JOIN rental r
    ON c.customer_id = r.customer_id
WHERE r.rental_date >= DATE_SUB(CURDATE(), INTERVAL 6 MONTH)
GROUP BY c.customer_id, c.first_name, c.last_name
HAVING COUNT(r.rental_id) > 10
ORDER BY rental_count DESC;
